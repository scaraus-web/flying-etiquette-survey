```python
import pandas as pd

# Load the CSV (even if extension is .xls, treat as CSV)
df = pd.read_csv("flying-etiquette.csv", encoding="utf-8", sep=",")

# 1. Remove duplicate rows
df = df.drop_duplicates()

# 2. Handle missing values
# Fill numeric columns with median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical/text columns with mode
for col in df.select_dtypes(include="object").columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# 3. Normalize numeric columns (Min-Max scaling 0–1)
for col in df.select_dtypes(include="number").columns:
    min_val, max_val = df[col].min(), df[col].max()
    if min_val != max_val:  # avoid division by zero
        df[col] = (df[col] - min_val) / (max_val - min_val)

# Save cleaned file
df.to_csv("flying-etiquette_clean.csv", index=False)

print("✅ Cleaning complete. Saved as 'flying-etiquette_clean.csv'")
```
